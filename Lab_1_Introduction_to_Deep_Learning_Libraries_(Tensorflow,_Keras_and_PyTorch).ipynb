# **TENSORFLOW**

An open-source deep learning framework developed by **Google** for building and training neural networks.

**Type:** Full end-to-end deep learning framework

**Developer:** Google Brain (2015)

**Core Features for Deep Learning:**

* Supports Dense, Convolutional (CNN), Recurrent (RNN/LSTM/GRU) layers

* **High-level APIs:** `tf.keras`, `tf.data for pipelines`, `tf.losses`, `tf.optimizers`

* **Distributed training:** tf.distribute.Strategy

* Mixed precision & TPU support for faster training

* Custom training loops using GradientTape

**Pros:**

* Excellent for large-scale models, distributed training

* Strong support for production deployment (TF Lite, TF Serving, TensorFlow.js)

* Large library of pretrained models and layers

**Cons:**

* Verbose for custom operations (before TF 2.x)

* Debugging static graphs is difficult

**Real-Life Deep Learning Usage:**

* CNNs for image recognition (Google Photos)

* NLP models (Google Translate)

**Example Use Case:**
Image classification, object detection, NLP models

```python
import tensorflow as tf
import numpy as np

# Step 1: Convert data to NumPy arrays
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 4, 6, 8], dtype=float)

# Step 2: Define model using Input layer (recommended)
model = tf.keras.Sequential([
    tf.keras.Input(shape=(1,)),  # preferred over input_shape in Dense
    tf.keras.layers.Dense(1)
])

# Step 3: Compile the model
model.compile(optimizer='sgd', loss='mean_squared_error')

# Step 4: Train the model
model.fit(x, y, epochs=10)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Convert lists to NumPy arrays
x = np.array([1, 2, 3, 4], dtype=float)
y = np.array([2, 4, 6, 8], dtype=float)

# Define the model
model = Sequential([
    Dense(1, input_shape=(1,))  # This works, or you can use tf.keras.Input(shape=(1,))
])

# Compile the model
model.compile(optimizer='sgd', loss='mse')

# Train the model
model.fit(x, y, epochs=10)
```

# **PyTorch**

An open-source deep learning framework developed by Facebook (Meta).

**Key Features:**

* Dynamic computation graph → graphs are created on the fly

* Pythonic & easy to debug → very developer-friendly

* Supports GPU acceleration and production deployment

* Widely used in research


```python
import torch
import torch.nn as nn
import torch.optim as optim

# Define a simple linear model
model = nn.Linear(1, 1)

# Loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Dummy data
x = torch.tensor([[1.],[2.],[3.],[4.]])
y = torch.tensor([[2.],[4.],[6.],[8.]])

# Training loop
for epoch in range(10):
    optimizer.zero_grad()
    outputs = model(x)
    loss = criterion(outputs, y)
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}, Loss: {loss.item()}")
```

# **Given Code demonstrates how input data is converted into tensors, passed through a neural network layer, and produces an output tensor using learned weight and bias tensors.**

```python
import tensorflow as tf                                                   # STEP 1 IMPORT TENSORFLOW
```

```python
# one data sample with 2 features                                         # STEP 2 INPUT DATA
x = tf.constant([[20.0, 80.0]])   # shape: (1, 2)
```

`20.0` and` 80.0 `are numerical features

`[[ ]] means:` 1 data sample, 2 input features

**Tensor details:**

**Type:** Tensor

**Shape:** (1, 2)

1 row (sample)

2 columns (features)

```python
model = tf.keras.Sequential([                                             # STEP 3 Create a Small Nueral Network
    tf.keras.layers.Dense(1, input_shape=(2,))   # Dense(1) → one neuron                                          # activation='relu',
])                                               #Input tensor shape → (2,)
```

`Sequential →` layers are arranged one after another

`Dense(1) → `one neuron

`input_shape=(2,) →` neuron expects 2 input values

**TensorFlow automatically creates:**

`Weight tensor` → shape (2, 1)

`Bias tensor` → shape (1,)

```python
y = model(x)                                                              # STEP 4 FORWARD PASS
print(y)        #  input tensor → weight tensor → bias tensor → output tensor(y)
```

1. Input tensor x enters the neuron

2. Each input is multiplied by a weight

3. All values are added

4. Bias is added

```python
weights, bias = model.layers[0].get_weights()
```

`model.layers[0]`→ first (and only) layer

`get_weights()` → returns: weight tensor and bias tensor

**Shapes:**

`weights →` (2, 1)

`bias → `(1,)

These tensors are learned parameters.

```python
print("Weights tensor:", weights)                                         # STEP 5 TENSOR VISUALIZATION
print("Bias tensor:", bias)
print("Output tensor:", y)
```

**Neuron formula:**

`Output = (x1 * w1)  + (x2 * w2) + bias`

# **TENSORFLOW WITH KERAS (DL PIPELINE)**
**“Binary Classification of Single/Multi-Sample Inputs Using TensorFlow/Keras Sequential Model”**

* Demonstrates building a simple feedforward neural network.

* Uses Dense layers with ReLU hidden layer and Sigmoid output layer.

* Performs forward pass, loss computation, training (backprop), and evaluation.

* Suitable for binary classification tasks with small input samples.

```python
import tensorflow as tf
# Step 1: Input Data
X = tf.constant([[20.0, 80.0]])       # 1 sample, 2 features
y = tf.constant([[1.0]])              # Binary label
```

```python
# Step 2: Model Design
tf_model = tf.keras.Sequential([
    tf.keras.layers.Dense(16, activation='relu', input_shape=(2,)),   # Hidden layer
    tf.keras.layers.Dense(1, activation='sigmoid')                    # Output layer
])
```

```python
# Step 3: Model Compilation
tf_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Step 4: Model Training
tf_model.fit(X, y, epochs=3, verbose=1)
```

```python
# Step 5: Model Evaluation
loss, accuracy = tf_model.evaluate(X, y)
print("TensorFlow/Keras Accuracy:", accuracy)
```

# **PYTORCH (DL PIPELINE)**

**“Binary Classification of Single/Multi-Sample Inputs Using PyTorch Dynamic Neural Network”**


* Demonstrates building a simple neural network class (nn.Module).

* Uses manual forward pass, ReLU hidden layer, Sigmoid output, BCELoss, and Adam optimizer.

* Shows training loop, gradient computation, weight updates, and evaluation explicitly.

* Ideal for understanding dynamic computation graph in PyTorch.

```python
import torch                        # torch → main PyTorch library, handles tensors.
import torch.nn as nn               # torch.nn → contains modules for neural network layers.
import torch.optim as optim         # torch.optim → contains optimizers like Adam, SGD.

# Step 1: Input Data
# --------------------------
X = torch.tensor([[20.0, 80.0]])  # X → input tensor (1 sample, 2 features: 20 & 80).
y = torch.tensor([[1.0]])         # y → target/output tensor (1 sample, binary label 1).
```

```python
# Step 2: Model Design
class SimpleNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 16)    # Hidden layer: 2 inputs → 16 neurons
        self.fc2 = nn.Linear(16, 1)    # Output layer: 16 → 1 neuron

    def forward(self, x):
        x = torch.relu(self.fc1(x))     # Apply ReLU activation to hidden layer
        x = torch.sigmoid(self.fc2(x))  # Apply Sigmoid activation to output layer
        return x

model = SimpleNN()
```

```python
# Step 3: Loss and Optimizer
criterion = nn.BCELoss()                            # nn.BCELoss() → Binary Cross-Entropy loss, suitable for binary classification.
optimizer = optim.Adam(model.parameters(), lr=0.01) # Adam optimizer; updates model weights automatically.
```

```python
# Step 4: Training Loop
epochs = 3
for epoch in range(epochs):
    optimizer.zero_grad()          # Reset gradients before backprop
    y_pred = model(X)              # Forward pass: compute predicted output
    loss = criterion(y_pred, y)    # Compute loss between predicted and target
    loss.backward()                # Backpropagation: compute gradients
    optimizer.step()               # Update weights using gradients
    print(f"Epoch {epoch+1}, Loss: {loss.item():.4f}")
```

`epochs `= 3 → Number of times we train over the same input.

`for epoch in range(epochs) →` Loop over each epoch.

`optimizer.zero_grad() →` PyTorch accumulates gradients, so we reset them each iteration.

`y_pred = model(X) → `Forward pass through the network; calls forward() internally.

`loss = criterion(y_pred, y) →` Compute how far predictions are from true labels.

`loss.backward() →` Backpropagation: calculates gradients of loss w.r.t each parameter.

`optimizer.step() →` Adjusts the weights using the gradients.

`print(...) →` Shows the loss after each epoch.

```python
# Step 5: Evaluation
with torch.no_grad():
    y_pred = model(X)
    prediction = (y_pred > 0.5).float()
    accuracy = (prediction == y).sum().item()
    print("PyTorch Prediction:", prediction.item())
    print("PyTorch Accuracy:", accuracy)
```

`with torch.no_grad(): → `Disables gradient tracking (we don’t need gradients during evaluation).

`y_pred = model(X) →` Compute predictions for input.

`prediction = (y_pred > 0.5).float() → `Threshold sigmoid output: >0.5 → 1, else 0.

`accuracy = (prediction == y).sum().item() → `Count how many predictions match target.

`prediction.item() →` Returns the predicted value as a Python scalar.

Prints predicted value and accuracy.

# **TENSORFLOW WITHOUT KERAS**

```python
import tensorflow as tf

# --------------------------
# Step 1: Input Data
# --------------------------
X = tf.constant([[20.0, 80.0]])
y = tf.constant([[1.0]])

# --------------------------
# Step 2: Initialize weights and bias
# --------------------------
w1 = tf.Variable(tf.random.uniform((2,16), -0.5, 0.5))
b1 = tf.Variable(tf.zeros((16,)))
w2 = tf.Variable(tf.random.uniform((16,1), -0.5, 0.5))
b2 = tf.Variable(tf.zeros((1,)))

# Learning rate
lr = 0.01
epochs = 3

# --------------------------
# Step 3: Training Loop
# --------------------------
for epoch in range(epochs):
    with tf.GradientTape() as tape:
        # Forward pass
        hidden = tf.nn.relu(tf.matmul(X, w1) + b1)
        output = tf.sigmoid(tf.matmul(hidden, w2) + b2)
        # Binary cross-entropy loss
        loss = -y*tf.math.log(output) - (1-y)*tf.math.log(1-output)
        loss = tf.reduce_mean(loss)

    # Compute gradients
    grads = tape.gradient(loss, [w1, b1, w2, b2])

    # Update weights
    w1.assign_sub(lr * grads[0])
    b1.assign_sub(lr * grads[1])
    w2.assign_sub(lr * grads[2])
    b2.assign_sub(lr * grads[3])

    print(f"Epoch {epoch+1}, Loss: {loss.numpy():.4f}")

# --------------------------
# Step 4: Evaluation
# --------------------------
hidden = tf.nn.relu(tf.matmul(X, w1) + b1)
output = tf.sigmoid(tf.matmul(hidden, w2) + b2)
prediction = tf.cast(output > 0.5, tf.float32)
print("TensorFlow Low-Level Prediction:", prediction.numpy())
```
